merge sort

```python
def chia(arr):
    if len(arr)<=1:
        return arr
    left=chia(arr[:len(arr)//2])
    right=chia(arr[len(arr)//2:])
    return merge(left,right)
def merge(left,right):
    res=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res
def binary_search(arr,key):
    left=0
    right=len(arr)-1
    mid=(left+right)//2
    while left<=right:
        if key==arr[mid]:
            return mid
        elif key<arr[mid]:
            right=mid-1
        else:
            left=mid+1
    return -1
my_list=[int(i) for i in input().split()]
print(chia(my_list))
key=int(input())
print(binary_search(chia(my_list),key))
# [0, 5, 2, 10, 4, 6]
# 2
```

quick sort

```python
def tim_vi_tri_pivot(arr,low,high):
    pivot=arr[high]
    i=low-1
    for j in range(low,high):
        if arr[j]<=pivot:
            i+=1
            arr[i],arr[j]=arr[j],arr[i]
    arr[i+1],arr[high]=arr[high],arr[i+1]
    return i+1
def quick_sort(arr,low,high):
    if low < high:
        p=tim_vi_tri_pivot(arr,low,high)
        quick_sort(arr,low,p-1)
        quick_sort(arr,p+1,high)
my_list=[int(i) for i in input().split()]
quick_sort(my_list,0,len(my_list)-1)
print(my_list)
    
```

```python
def binary_search(arr,left,right,target):
    mid=(left+right)//2
    while left<=right:
        if arr[mid]==target:
            return mid
        elif arr[mid]>target:
            return binary_search(arr,left,mid-1,target)
        else: return binary_search(arr,mid+1,right,target)
    return -1
arr=[int(i) for i in input().split()]
target=int(input())
print(binary_search(arr,0,len(arr)-1,target))
```

thời gian nào như vậy, hãy

solve in < O(m.n), hint: 
binary search in the 
queries list

```python
def merge(l,r,check):
    res=[]
    i=0
    j=0
    while i<len(l) and j<len(r):
        if l[i]<r[j]:
            if check:
                res.append(l[i])
                i+=1
            else:
                res.append(r[j])
                j+=1
        else:
            if check:

                res.append(r[j])
                j+=1
            else:
                res.append(l[i])
                i+=1
    res.extend(l[i:])
    res.extend(r[j:])
    return res
a=[int(i) for i in input().split()]
b=[int(i) for i in input().split()]
if a[0]<a[-1] and b[0]<b[-1]:
    res=merge(a,b,1)
else:
    res=merge(a,b,0)
for i in res:
        print(i,end=' ')
    
```

W7A29 
Tìm Dãy con tăng dần liên tục dài nhất 
của một danh sách nhập vào từ bàn 
phím. Lấy dãy đầu tiên nếu có nhiều dãy 
có số phần tử bằng nhau 

```python
a=[int(i) for i in input().split()]
len_max=1
idx_max=0
len_cur=1
idx_cur=0
for i in range(1,len(a)):
    if a[i]>=a[i-1]:
        len_cur+=1
    else:
        idx_cur=i
        len_cur=1
    if len_cur>len_max:
        idx_max=idx_cur
        len_max=len_cur
res=a[idx_max:idx_max+len_max]
for i in res:
    print(i,end=' ')
```

W7A30 
Tìm Dãy con tăng dần KHÔNG liên tục 
dài nhất của một danh sách nhập vào từ 
bàn phím  

```python
a=[int(i) for i in input().split()]
l=[1]*(len(a)) # mang luu do dai dai nhat ket thuc tai vi tri i
r=[-1]*len(a) # mang luu idx cua so trc no de truy vet
for i in range(1,len(a)):
    for j in range(i):
        if a[i]>=a[j] and l[i]<l[j]+1:
            l[i]=l[j]+1
            r[i]=j
len_max=max(l)
idx=l.index(len_max)
res=[]
while idx!=-1:
    res.append(a[idx])
    idx=r[idx]
res.reverse()
for i in res:
    print(i,end=' ')

```
