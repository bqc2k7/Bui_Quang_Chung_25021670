merge sort

```python
def chia(arr):
    if len(arr)<=1:
        return arr
    left=chia(arr[:len(arr)//2])
    right=chia(arr[len(arr)//2:])
    return merge(left,right)
def merge(left,right):
    res=[]
    i=0
    j=0
    while i<len(left) and j<len(right):
        if left[i]<right[j]:
            res.append(left[i])
            i+=1
        else:
            res.append(right[j])
            j+=1
    res.extend(left[i:])
    res.extend(right[j:])
    return res
def binary_search(arr,key):
    left=0
    right=len(arr)-1
    mid=(left+right)//2
    while left<=right:
        if key==arr[mid]:
            return mid
        elif key<arr[mid]:
            right=mid-1
        else:
            left=mid+1
    return -1
my_list=[int(i) for i in input().split()]
print(chia(my_list))
key=int(input())
print(binary_search(chia(my_list),key))
# [0, 5, 2, 10, 4, 6]
# 2
```

quick sort

```python
def tim_vi_tri_pivot(arr,low,high):
    pivot=arr[high]
    i=low-1
    for j in range(low,high):
        if arr[j]<=pivot:
            i+=1
            arr[i],arr[j]=arr[j],arr[i]
    arr[i+1],arr[high]=arr[high],arr[i+1]
    return i+1
def quick_sort(arr,low,high):
    if low < high:
        p=tim_vi_tri_pivot(arr,low,high)
        quick_sort(arr,low,p-1)
        quick_sort(arr,p+1,high)
my_list=[int(i) for i in input().split()]
quick_sort(my_list,0,len(my_list)-1)
print(my_list)
    
```
