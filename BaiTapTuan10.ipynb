Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ 
thuật toán sắp xếp nào 

```python
def merge(l,r):
    res=[]
    i=0
    j=0
    while i<len(l) and j<len(r):
        if l[i]<r[j]:
            res.append(l[i])
            i+=1
        else:
            res.append(r[j])
            j+=1
    res.extend(l[i:])
    res.extend(r[j:])
    return res
a=[int(i) for i in input().split()]
b=[int(i) for i in input().split()]
print(merge(a,b))
```

có thuộc list2 hay không (có phải là dãy con liên tiếp của list2 hay

```python
list1=[int(i) for i in input().split()]
list2=[int(i) for i in input().split()]
if len(list1)>len(list2):
    print("NO")
else:
    i=0
    j=0
    check=False
    while i<len(list1) and j<len(list2):
        if list1[i]==list2[j]:
            x=j+len(list1)-1
            if x<len(list2):
                tmp=list2[j:x+1]
                if tmp==list1:
                    check=True
                    break
                else:
                    j+=1
            else:
                check=False
                break
        else:
            j+=1
    if check:
        print("YES")
    else:
        print("NO")
```

Đầu ra: In ra màn hình hoán vị tiếp theo của hoán vị đã cho. 

```python
n=int(input())
a=[int(i) for i in input().split()]
flag=False
for i in range(n-2,-1,-1):
    if a[i]<a[i+1]:
        k=i
        flag=True
        break
if not flag:
    for i in range(n-1,-1,-1):
        print(a[i],end=' ')
else:
    ans=a[:k]
    idx=0
    for i in range(k+1,len(a)):
        if a[i]>a[k]:
            idx=i
    a[idx],a[k]=a[k],a[idx]
    ans.append(a[k])
    tmp=a[k+1:]
    tmp.reverse()
    ans.extend(tmp)
    for i in ans:
        print(i,end=' ')

```

Hàm trả về `True` nếu trong quá trình di chuyển, rô-bốt đi vào một ô mà nó đã từng đi qua trước đó.

Ràng buộc: 
1. Chỉ được sử dụng tối đa 6 phép so sánh (ví dụ: <, >, <=, >=). 
2. KHÔNG ĐƯỢC DÙNG bất kỳ hàm sắp xếp nào (như 
`sorted()`, `list.sort()`). 
 
Chú ý: Bạn chỉ cần viết hàm `Median` theo yêu cầu.  
Phần đọc input và gọi hàm (hàm main) đã có sẵn. 

```python
nums=[int(i) for i in input().split()]
def median(nums):
    a,b,c,d,e=nums[0],nums[1],nums[2],nums[3],nums[4]
    if a>b:
        a,b=b,a
    if c>d:
        c,d=d,c
    if a>c:
        a,b,c,d=c,d,a,b
    a=e
    if a>b:
        a,b=b,a
    if a>c:
        a,b,c,d=c,d,a,b
    if b<c:
        return b
    else:
        return c
print(median(nums))
    
```

[Tìm bán kính lò sưởi nhỏ nhất] 
 
Mùa đông đang đến! Bạn được cung cấp vị trí của các ngôi nhà 
và các lò sưởi trên một đường thẳng. 
Nhiệm vụ của bạn là tìm bán kính r nhỏ nhất cho các lò sưởi sao 
cho mọi ngôi nhà đều nằm trong phạm vi sưởi ấm của ít nhất một 
lò sưởi. 
 
Tất cả các lò sưởi đều có cùng một bán kính r. 
Một ngôi nhà tại vị trí h được sưởi ấm bởi một lò sưởi tại vị trí t 
nếu |h - t| <= r. 
 
Viết hàm `def findRadius(houses, heaters):` 
Hàm nhận đầu vào là hai danh sách (list) các số nguyên: - `houses`: chứa vị trí của các ngôi nhà. - `heaters`: chứa vị trí của các lò sưởi. 
Hàm trả về một số nguyên là bán kính r nhỏ nhất cần thiết. 
 
Lưu ý: - Số lượng nhà và lò sưởi là một số không âm và không vượt quá 
25000. - Vị trí của các ngôi nhà và lò sưởi là một số không âm và không 
vượt quá 10^9.

```python
def tim_lo_gan_nhat(a,x):
    l=0
    r=len(a)-1
    while l<=r:
        m=(l+r)//2
        if a[m]==x:
            return 0
        elif a[m]>x:
            r=m-1
        else:
            l=m+1
    d1,d2=0,0
    if r>=0:
        d1=abs(x-a[r])
    else:
        d1=float('inf')
    if l<len(a):
        d2=abs(x-a[l])
    else:
        d2=float('inf')
    return min(d1,d2)
houses=[int(i) for i in input().split()]
heaters=[int(i) for i in input().split()]
houses.sort()
heaters.sort()
ans=0
for i in houses:
    d=tim_lo_gan_nhat(heaters,i)
    ans=max(ans,d)
print(ans)
```
