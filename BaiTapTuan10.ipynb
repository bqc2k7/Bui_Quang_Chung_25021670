W9A1 
Viết hàm merge hai danh sách đã sắp xếp mà không dùng bất cứ 
thuật toán sắp xếp nào 

```python
def merge(l,r):
    res=[]
    i=0
    j=0
    while i<len(l) and j<len(r):
        if l[i]<r[j]:
            res.append(l[i])
            i+=1
        else:
            res.append(r[j])
            j+=1
    res.extend(l[i:])
    res.extend(r[j:])
    return res
a=[int(i) for i in input().split()]
b=[int(i) for i in input().split()]
print(merge(a,b))
```

W9A3 
Cho một số nguyên dương, hãy thực hiện hoán đổi đúng 2 chữ 
số để được một số nguyên nhỏ nhất có thể

```python
n=input()
ans=['9']*len(n)
for i in range(len(n)):
    for j in range(i+1,len(n)):
        tmp=list(n)
        tmp[i],tmp[j]=tmp[j],tmp[i]
        ans=min(ans,tmp)
if ans[0]=='0':
    x=1
    while ans[x]=='0':
        x+=1
    for i in range(x,len(ans)):
        print(ans[i],end='')
else:
    for i in ans:
        print(i,end='')
```

W9A5 
Cho hai mảng số nguyên list1 và list2, viết hàm kiểm tra xem list1 
có thuộc list2 hay không (có phải là dãy con liên tiếp của list2 hay 
không)?

```python
list1=[int(i) for i in input().split()]
list2=[int(i) for i in input().split()]
if len(list1)>len(list2):
    print("NO")
else:
    i=0
    j=0
    check=False
    while i<len(list1) and j<len(list2):
        if list1[i]==list2[j]:
            x=j+len(list1)-1
            if x<len(list2):
                tmp=list2[j:x+1]
                if tmp==list1:
                    check=True
                    break
                else:
                    j+=1
            else:
                check=False
                break
        else:
            j+=1
    if check:
        print("YES")
    else:
        print("NO")
```

```python
def sinh_tap_con(nums):
    ket_qua = []
    
    # Hàm đệ quy backtrack
    # index: vị trí phần tử đang xét trong nums
    # tap_hien_tai: danh sách tạm thời chứa các số đã chọn
    def backtrack(start_index, tap_hien_tai):
        
        # 1. Ghi nhận kết quả
        # Lưu bản sao (copy) của tập hiện tại vào kết quả
        # Lưu ý: phải dùng tap_hien_tai[:] để copy, nếu không sẽ bị lỗi tham chiếu
        ket_qua.append(tap_hien_tai[:])
        
        # 2. Duyệt qua các lựa chọn
        for i in range(start_index, len(nums)):
            # --- BƯỚC THỬ (CHỌN) ---
            tap_hien_tai.append(nums[i]) 
            
            # --- BƯỚC ĐỆ QUY (ĐI TIẾP) ---
            # Gọi đệ quy với i + 1 để xét các số phía sau
            backtrack(i + 1, tap_hien_tai)
            
            # --- BƯỚC QUAY LUI (TRẢ LẠI) ---
            # Xóa số vừa thêm để quay về trạng thái cũ -> thử số khác
            tap_hien_tai.pop()

    # Bắt đầu gọi hàm từ vị trí 0
    backtrack(0, [])
    return ket_qua

# Chạy thử
nums = [1, 2, 3]
print(sinh_tap_con(nums))
```
