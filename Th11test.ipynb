Bạn hãy thiết kế một lớp đối tượng có tên là ComplexNumber để biểu diễn số phức. Số phức có dạng:
z = a + bi
trong đó a là phần thực, b là phần ảo, và i^2 = -1.

Sử dụng “ghi đè” (override) các phương thức đặc biệt sau để toán tử Python hoạt động đúng:

- __init__(self, initRealPart, initImaginaryPart):
- Lưu phần thực và phần ảo (float)

- __add__(self, other):
- Cài đặt phép cộng bằng toán tử +.
Trả về một đối tượng ComplexNumber MỚI.

- __mul__(self, other):
- Cài đặt phép nhân bằng toán tử *.
Trả về một đối tượng ComplexNumber MỚI.

- __abs__(self):
- Cài đặt abs(z) để trả về mô-đun của số phức:
- Trả về float.

- __str__(self):
- Trả về chuỗi biểu diễn số phức để in/kiểm tra,
ví dụ: "a+bi"nếu b>0 hoặc "a-bi" nếu b<0.

- conjugate(self):
- Trả về số phức liên hợp: (một đối tượng ComplexNumber MỚI).


RÀNG BUỘC KIỂM TRA:
Trong chương trình chính, bắt buộc phải dùng các biểu thức sau:
- z1 + z2
- z1 * z2
- abs(z1 * z2)

--------------------------------------------------------------------

DỮ LIỆU ĐẦU VÀO (INPUT)
- Gồm 2 dòng:
- Dòng 1: Hai số thực a1 b1 cách nhau bởi dấu cách (phần thực và phần ảo của số phức thứ nhất).
- Dòng 2: Hai số thực a2 b2 cách nhau bởi dấu cách (phần thực và phần ảo của số phức thứ hai).
- Điều kiện: Các giá trị tuyệt đối không vượt quá 100.

--------------------------------------------------------------------

DỮ LIỆU ĐẦU RA (OUTPUT)
In ra 4 dòng theo thứ tự sau (mỗi giá trị làm tròn đến 2 chữ số thập phân):
- Dòng 1: Tổng z1 + z2
- Dòng 2: Tích z1 * z2
- Dòng 3: Mô-đun của tích |z1 * z2|
- Dòng 4: Số phức liên hợp của tích conj(z1 * z2)

```python
import math

class sophuc():
    def __init__(self, thuc, ao):
        self.thuc = float(thuc)
        self.ao = float(ao)

    def add(self, other):
        return sophuc(self.thuc + other.thuc, self.ao + other.ao)

    def mul(self, other):
        a = self.thuc * other.thuc - self.ao * other.ao
        b = self.thuc * other.ao + self.ao * other.thuc
        return sophuc(a, b)

    def abs(self): 
        return (self.thuc**2 + self.ao**2)**0.5

    def str(self,check):
        phan_thuc = f"{self.thuc:.2f}"
        phan_ao = f"{abs(self.ao):.2f}"
        
        if self.ao >= 0:
          if check and self.ao==0:
            return f"{phan_thuc} + -{phan_ao}i"
          return f"{phan_thuc} + {phan_ao}i"
        else:
            return f"{phan_thuc} - {phan_ao}i"

    def conj(self):
        return sophuc(self.thuc, -1 * self.ao)

a1, b1 = map(float, input().split())
z1 = sophuc(a1, b1)

a2, b2 = map(float, input().split())
z2 = sophuc(a2, b2)

z3 = z1.add(z2) 
print(f"{z3.str(0)}")

z4 = z1.mul(z2)
print(f"{z4.str(0)}")

print(f"{z4.abs():.2f}")

z4_lien_hop = z4.conj()
print(f"{z4_lien_hop.str(1)}")
```
