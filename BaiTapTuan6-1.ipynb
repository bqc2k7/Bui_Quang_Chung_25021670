Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo 
dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key

```python
s=input()
tach_k_v=s.split()
dt={}
for i in tach_k_v:
    idx=i.index(":")
    k=i[:idx]
    v=i[idx+1:]
    if k not in dt:
        dt[k]=[]
    if v not in dt[k]:
        dt[k].append(v)
for k,v in dt.items():
    print(k,v)
```

dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là 
một tuple, hai số trong tuple có thể giống nhau, thứ tự xuất hiện trong danh 
sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được 
sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
n=input()
k=int(input())
day_so=list(map(int,n.split()))
ans=[]
freq={}
for b in day_so:
    x=k-b
    if x in freq:
        for _ in range(freq[x]):
            ans.append((x,b))
    freq[b]=freq.get(b,0)+1
ans.reverse()
# ans.sort(key=lambda t: t[0])
print(ans)
```

Mấy bài làm thêm: Viết hàm kiểm tra một số có phải số Armstrong hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x


```python
def so_chu_so(x):
    cnt=0
    while(x>0):
        x//=10
        cnt+=1
    return cnt
def is_armstrong(x):
    tmp=x
    sum=0
    cnt=so_chu_so(x)
    while tmp>0:
        a=tmp%10
        sum+=(a**cnt)
        tmp//=10
    return sum==x
x=int(input())
# print(so_chu_so(x))
print(is_armstrong(x))
```

Viết hàm kiêểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x

VIết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.


```python
def is_doi_xung(a):
    ra=0
    tmp=a
    while tmp>0:
        ra=ra*10+tmp%10
        tmp//=10
    return ra==a
a=int(input())
print(is_doi_xung(a))
```

VIết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.
tổng số = tích sô

```python
def so_ki_dieu(x):
    tich=1
    tong=0
    while x>0:
        tich*=(x%10)
        tong+=x%10
        x//=10
    return tich==tong
x=int(input())
print(so_ki_dieu(x))
```
